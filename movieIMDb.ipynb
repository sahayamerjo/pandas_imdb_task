1. Which are the highest grossing movies within the data?

```python
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].str.replace('$','')
```

```python
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].str.replace('INR','')
```

```python
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].str.replace('PKR','')
```

```python
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].str.replace('NPR','')
```

```python
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].str.replace('GBP','')
```

```python
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].str.replace(',','')
```

```python
mov["worlwide_gross_income"].dtype
```

```python
mov = mov.dropna(subset=["worlwide_gross_income"])
mov["worlwide_gross_income"] = mov["worlwide_gross_income"].astype(int)
```

```python
mov["worlwide_gross_income"].dtype
```

```python
highest = mov.sort_values(by="worlwide_gross_income", ascending=False).iloc[0:20]
highest["worlwide_gross_income"]
```

2. Which movies have the highest average user vote?

```python
highest_avg_vote = mov.sort_values(by="avg_vote", ascending=False).iloc[0:20]
highest_avg_vote[["title", "avg_vote"]]
```

3. Which movies have the most 'polarised' votes?

```python
most_polarised_votes = mov.sort_values(by="votes", ascending=False).iloc[0:20]
most_polarised_votes[["title", "votes"]]
```

5. What is the gross income per director?

```python
director_gross = mov.groupby("director")["worlwide_gross_income"].sum().head(60)
director_gross
```

```python
highest_gross = 'INR ' + director_gross.sort_values(ascending=False).astype(str)
highest_gross
```
